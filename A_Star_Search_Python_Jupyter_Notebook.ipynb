# A* Search Algorithm in Python

This notebook implements A* search on a weighted graph using:

\[
f(n) = g(n) + h(n)
\]

- `g(n)`: actual cost from the start node
- `h(n)`: estimated cost from the current node to the goal
- `f(n)`: estimated total cost of a path through the current node


## 1. Define the weighted graph

Each node stores its neighbors and the corresponding edge costs. The graph is undirected, so every edge is represented in both directions.

```python
graph = {
    'S': {'A': 2, 'B': 5},
    'A': {'S': 2, 'C': 2, 'D': 5},
    'B': {'S': 5, 'D': 1},
    'C': {'A': 2, 'G': 5},
    'D': {'A': 5, 'B': 1, 'G': 3},
    'G': {'C': 5, 'D': 3}
}
```

## 2. Define heuristic values

These values estimate the remaining cost from each node to the goal node `G`.

```python
heuristic = {
    'S': 7,
    'A': 6,
    'B': 3,
    'C': 4,
    'D': 2,
    'G': 0
}
```

## 3. Define the A* search function

The priority queue selects the node with the smallest `f(n)`. If a cheaper route to a node is discovered, its `g_score` and `parent` are updated.

```python
import heapq

def a_star_search(graph, heuristic, start, goal):
    priority_queue = []
    g_score = {start: 0}
    parent = {start: None}
    traversal = []

    # Add the start node with f(n) = g(n) + h(n)
    start_f = g_score[start] + heuristic[start]
    heapq.heappush(priority_queue, (start_f, start))

    while priority_queue:
        f_value, current = heapq.heappop(priority_queue)
        current_g = g_score[current]

        # Ignore an outdated queue entry
        if f_value != current_g + heuristic[current]:
            continue

        traversal.append(current)
        print(
            f"Visiting: {current}, "
            f"g(n)={current_g}, "
            f"h(n)={heuristic[current]}, "
            f"f(n)={f_value}"
        )

        # Check whether the goal has been reached
        if current == goal:
            path = []
            node = goal

            # Reconstruct the path using parent pointers
            while node is not None:
                path.append(node)
                node = parent[node]

            path.reverse()
            return traversal, path, g_score[goal]

        # Explore neighboring nodes
        for neighbor, cost in graph[current].items():
            tentative_g = current_g + cost

            # Update only if a better path is found
            if tentative_g < g_score.get(neighbor, float('inf')):
                g_score[neighbor] = tentative_g
                parent[neighbor] = current

                f_neighbor = tentative_g + heuristic[neighbor]
                heapq.heappush(
                    priority_queue,
                    (f_neighbor, neighbor)
                )

    return traversal, None, float('inf')
```

## 4. Run A* search

Set the start and goal nodes, then call the function.

```python
start = 'S'
goal = 'G'

traversal, path, total_cost = a_star_search(
    graph, heuristic, start, goal
)
```

## 5. Display the result

```python
print("\nTraversal:", traversal)
print("Solution Path:", path)
print("Total Cost:", total_cost)
```

## Expected result

- Expansion order: `S, A, B, C, D, G`
- Returned optimal path: `S → A → C → G`
- Total path cost: **9**

There is another optimal path, `S → B → D → G`, which also has a cost of 9. With the supplied graph, heuristic values, and tie-breaking, this notebook returns `S → A → C → G`.
