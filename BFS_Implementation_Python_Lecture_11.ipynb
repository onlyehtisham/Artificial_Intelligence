# Breadth-First Search (BFS) in Python
## Lecture 11 — Practical Implementation

**Learning goals**
- Represent a graph using an adjacency list.
- Implement BFS with a FIFO queue.
- Prevent duplicate discovery with a visited set.
- Store parent links and reconstruct the solution path.
- Distinguish traversal order from solution path.

Neighbors are processed in **alphabetical order** for reproducible results.

## 1. Represent the graph

This is an undirected graph: each edge must appear in both endpoints' neighbor lists.

```python
graph = {
    'A': ['B', 'D'],
    'B': ['A', 'C', 'E'],
    'C': ['B', 'E', 'F', 'G'],
    'D': ['A', 'E'],
    'E': ['B', 'C', 'D', 'G'],
    'F': ['C', 'H', 'I'],
    'G': ['C', 'E', 'I'],
    'H': ['F', 'I'],
    'I': ['F', 'G', 'H']
}

for node, neighbors in graph.items():
    print(f"{node}: {neighbors}")
```

## 2. Queue basics

BFS uses a FIFO queue (First In, First Out). `deque` supports efficient insertion at the rear and removal from the front.

```python
from collections import deque

queue = deque(['A'])
print("Initial queue:", list(queue))

queue.append('B')
queue.append('D')
print("After adding B and D:", list(queue))

first_node = queue.popleft()
print("Removed from front:", first_node)
print("Queue now:", list(queue))
```

## 3. Complete BFS implementation

The goal is tested when a node is removed from the queue. When a new neighbor is discovered, we mark it visited, store its parent, and enqueue it.

```python
def bfs(graph, start, goal):
    queue = deque([start])
    visited = {start}
    parent = {start: None}
    traversal = []

    while queue:
        current = queue.popleft()
        traversal.append(current)

        if current == goal:
            break

        for neighbor in sorted(graph[current]):
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current
                queue.append(neighbor)

    if goal not in parent:
        return traversal, None, parent

    # Follow parent pointers backward, then reverse the path
    path = []
    current = goal
    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()
    return traversal, path, parent
```

## 4. Run BFS from A to H

```python
traversal, path, parent = bfs(graph, 'A', 'H')

print("BFS traversal order:", traversal)
print("BFS solution path:", path)
print("Parent dictionary:", parent)
```

## 5. Trace the queue step by step

This version prints the queue, visited set, and newly discovered neighbors after each node is processed.

```python
def bfs_trace(graph, start, goal):
    queue = deque([start])
    visited = {start}
    parent = {start: None}
    traversal = []

    print(f"Start: queue={list(queue)}, visited={sorted(visited)}")

    while queue:
        current = queue.popleft()
        traversal.append(current)

        if current == goal:
            print(f"Process {current}: GOAL FOUND; queue={list(queue)}")
            break

        newly_added = []
        for neighbor in sorted(graph[current]):
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current
                queue.append(neighbor)
                newly_added.append(neighbor)

        print(
            f"Process {current}: added={newly_added}, "
            f"queue={list(queue)}, visited={sorted(visited)}"
        )

    if goal not in parent:
        path = None
    else:
        path = []
        node = goal
        while node is not None:
            path.append(node)
            node = parent[node]
        path.reverse()

    print("\nTraversal:", traversal)
    print("Solution path:", path)
    print("Parents:", parent)

bfs_trace(graph, 'A', 'H')
```

## 6. Try other start and goal nodes

Change `start` and `goal` to experiment.

```python
start = 'A'
goal = 'G'

traversal, path, parent = bfs(graph, start, goal)
print("Start:", start)
print("Goal:", goal)
print("Traversal:", traversal)
print("Solution path:", path)
```

## 7. Unreachable goal

If the goal is not reachable from the start, the function returns `None` for the solution path.

```python
traversal, path, parent = bfs(graph, 'A', 'Z')
print("Traversal:", traversal)
print("Solution path:", path)  # None means no solution was found
```

## Key takeaways

- **Queue:** controls processing order (FIFO).
- **Visited set:** prevents duplicate discovery.
- **Parent dictionary:** records which node first discovered each node.
- **Traversal:** records processing order.
- **Solution path:** follow parent pointers backward from the goal, then reverse.
- **Complexity:** `O(V + E)` time and `O(V)` auxiliary space for an adjacency-list graph, excluding graph storage.
