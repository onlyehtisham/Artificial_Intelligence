# Depth-First Search (DFS) in Python — Implementation with Goal Search

**Lecture 13 | Artificial Intelligence**

This notebook implements iterative DFS using an adjacency list, a stack, a visited set, a parent dictionary, and a traversal list. Nodes are marked visited **when pushed onto the stack**.

We search from start node `A` to goal node `I` using the graph from the lecture.


## 1. Graph representation

The graph is undirected, so every edge appears in both endpoints' neighbor lists. Neighbor order affects the DFS traversal order.


````python
graph = {
    'A': ['B', 'D'],
    'B': ['A', 'C', 'E'],
    'C': ['B', 'E', 'F', 'G'],
    'D': ['A', 'E'],
    'E': ['B', 'C', 'D', 'G'],
    'F': ['C', 'H', 'I'],
    'G': ['C', 'E', 'I'],
    'H': ['F', 'I'],
    'I': ['F', 'G', 'H']
}

for node, neighbors in graph.items():
    print(f'{node}: {neighbors}')
````

## 2. The four important DFS variables

- **`stack`**: nodes waiting to be processed; the rightmost item is the top (LIFO).
- **`visited`**: nodes already discovered and pushed onto the stack.
- **`parent`**: records which node first discovered each node; used to reconstruct the solution path.
- **`traversal`**: records the order in which nodes are popped and processed.

The starting node has `parent[start] = None` because it has no predecessor.


## 3. Complete DFS function

The function accepts `graph`, `start`, and `goal`, and returns both the traversal order and the solution path. The goal is tested when a node is popped, so the traversal includes the goal when it is found.


````python
def dfs(graph, start, goal):
    stack = [start]
    visited = {start}
    parent = {start: None}
    traversal = []

    while stack:
        current = stack.pop()
        traversal.append(current)

        # Stop when the goal is popped and processed
        if current == goal:
            break

        # Reverse neighbors so the first listed neighbor is processed first
        for neighbor in reversed(graph[current]):
            if neighbor not in visited:
                visited.add(neighbor)       # mark visited on push
                parent[neighbor] = current  # remember predecessor
                stack.append(neighbor)

    # Reconstruct the path from goal back to start, if found
    path = []
    if traversal and traversal[-1] == goal:
        node = goal
        while node is not None:
            path.append(node)
            node = parent[node]
        path.reverse()

    return traversal, path
````

## 4. Run DFS from A to I

Because we stop as soon as `I` is popped, nodes that remain on the stack are not processed after the goal is found.


````python
start = 'A'
goal = 'I'

traversal, path = dfs(graph, start, goal)

print('DFS Traversal Order:', traversal)
print(f'Path from {start} to {goal}:', path)
````

### Expected output

```text
DFS Traversal Order: ['A', 'B', 'C', 'F', 'H', 'I']
Path from A to I: ['A', 'B', 'C', 'F', 'I']
```

The traversal includes `H` because DFS explores it before `I` under this neighbor order. The solution path does not include `H`: the parent dictionary shows that `I` was first discovered from `F`.


## 5. Trace the stack, visited set, parent dictionary, and traversal list

The stack is printed from bottom to top, with the rightmost item at the top.


````python
def dfs_trace(graph, start, goal):
    stack = [start]
    visited = {start}
    parent = {start: None}
    traversal = []
    step = 0

    print(f'Start: stack={stack}, visited={visited}, parent={parent}')

    while stack:
        current = stack.pop()
        traversal.append(current)
        print(f'\\nStep {step + 1}: pop {current}')

        if current == goal:
            print('Goal found!')
            break

        pushed = []
        for neighbor in reversed(graph[current]):
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current
                stack.append(neighbor)
                pushed.append(neighbor)

        print('New nodes pushed (in push order):', pushed)
        print('Stack (bottom -> top):', stack)
        print('Visited set:', sorted(visited))
        print('Parent dictionary:', parent)
        print('Traversal list:', traversal)
        step += 1

    path = []
    if traversal and traversal[-1] == goal:
        node = goal
        while node is not None:
            path.append(node)
            node = parent[node]
        path.reverse()

    print('\\nFinal traversal:', traversal)
    print('Final path:', path)
    return traversal, path

trace_traversal, trace_path = dfs_trace(graph, 'A', 'I')
````

## 6. Try another start node or goal

Change `start` and `goal` and rerun this cell. Traversal and path may change depending on the starting node and neighbor order.


````python
start = 'D'
goal = 'H'

traversal, path = dfs(graph, start, goal)
print('DFS Traversal Order:', traversal)
print(f'Path from {start} to {goal}:', path)
````

## 7. Complexity analysis

For an adjacency-list graph implementation:

- **Time complexity:** `O(V + E)`
- **Space complexity:** `O(V)`

Here, `V` is the number of vertices and `E` is the number of edges. The stack, visited set, parent dictionary, and traversal list require at most linear space in the number of vertices.
