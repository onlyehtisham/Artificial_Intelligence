## Core idea

GBFS selects the frontier node with the smallest estimated cost to the goal:

\[
f(n) = h(n)
\]

The heuristic `h(n)` estimates the remaining cost. GBFS does not include the path cost already travelled.

## 1. Import the required module
`heapq` provides a min-heap: the smallest priority is removed first.

```python
import heapq
```

## 2. Graph representation
We use the same undirected A–I graph from the previous search lectures.

```python
graph = {
    'A': ['B', 'D'],
    'B': ['A', 'C', 'E'],
    'C': ['B', 'E', 'F', 'G'],
    'D': ['A', 'E'],
    'E': ['B', 'C', 'D', 'G'],
    'F': ['C', 'H', 'I'],
    'G': ['C', 'E', 'I'],
    'H': ['F', 'I'],
    'I': ['F', 'G', 'H']
}

for node, neighbors in graph.items():
    print(f"{node}: {neighbors}")
```

## 3. Heuristic dictionary
Goal node is I. Values are illustrative estimates of remaining cost, not edge costs. The goal has value zero.

```python
heuristic = {
    'A': 7, 'B': 6, 'C': 4,
    'D': 7, 'E': 3, 'F': 1,
    'G': 2, 'H': 1, 'I': 0
}

for node, value in heuristic.items():
    print(f"h({node}) = {value}")
```

## 4. Understanding the priority queue
Each entry is `(heuristic_value, node)`. The smallest heuristic value is selected first.

```python
priority_queue = []
heapq.heappush(priority_queue, (6, 'B'))
heapq.heappush(priority_queue, (4, 'C'))
heapq.heappush(priority_queue, (3, 'E'))

print("Heap:", priority_queue)
print("Removed first:", heapq.heappop(priority_queue))
```

## 5. Four important data structures

| Variable | Type | Purpose |
|---|---|---|
| `priority_queue` | list managed by `heapq` | Stores frontier nodes by heuristic priority |
| `visited` | set | Tracks nodes already explored |
| `parent` | dictionary | Records how each discovered node was reached |
| `traversal` | list | Records exploration order |

## 6. Complete GBFS implementation

```python
def greedy_best_first_search(graph, heuristic, start, goal):
    # Priority queue stores (heuristic value, node)
    priority_queue = []

    # Track explored nodes
    visited = set()

    # Track the parent of each discovered node
    parent = {start: None}

    # Record exploration order
    traversal = []

    # Insert the starting node
    heapq.heappush(priority_queue, (heuristic[start], start))

    while priority_queue:
        # Remove the node with the smallest heuristic value
        h_value, current = heapq.heappop(priority_queue)

        # Skip nodes already explored
        if current in visited:
            continue

        visited.add(current)
        traversal.append(current)

        print(f"Visiting: {current}, h(n) = {h_value}")

        # Stop when the goal is selected
        if current == goal:
            print("\nGoal found!")

            # Reconstruct the solution path
            path = []
            node = goal

            while node is not None:
                path.append(node)
                node = parent[node]

            path.reverse()
            print("Traversal:", traversal)
            print("Solution Path:", path)
            return traversal, path

        # Add neighbors that have not already been discovered
        for neighbor in graph[current]:
            if neighbor not in visited and neighbor not in parent:
                parent[neighbor] = current
                heapq.heappush(
                    priority_queue,
                    (heuristic[neighbor], neighbor)
                )

    print("Goal not found.")
    return traversal, None
```

## 7. Run the algorithm
Search from A to I.

```python
start = 'A'
goal = 'I'

traversal, path = greedy_best_first_search(
    graph, heuristic, start, goal
)
```

## 8. Expected result

- **Traversal:** `['A', 'B', 'E', 'G', 'I']`
- **Solution path:** `['A', 'B', 'E', 'G', 'I']`

B is selected before D, E before C, then G, and finally I. In this example the traversal and solution path happen to be identical; they are not necessarily identical in every graph.

## 9. Priority queue trace

Entries are shown as `(h(n), node)`. The table is sorted by priority for readability; the internal heap list is not necessarily fully sorted.

| Selected node | h(n) | Queue after expansion |

| B | 6 | (3, E), (4, C), (7, D) |
| E | 3 | (2, G), (4, C), (7, D) |
| G | 2 | (0, I), (4, C), (7, D) |
| I | 0 | Goal selected — stop |

## Key takeaways
- GBFS is an informed search algorithm.
- It uses `f(n) = h(n)` to prioritize frontier nodes.
- `heapq` implements the min-priority queue.
- `visited` prevents repeated expansion.
- `parent` supports path reconstruction.
- `traversal` records exploration order.
- GBFS is not guaranteed to find the lowest-cost path because it ignores the cost already incurred.
